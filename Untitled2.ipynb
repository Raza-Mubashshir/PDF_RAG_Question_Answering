```python
import torch
from pypdf import PdfReader
from sentence_transformers import SentenceTransformer, util
from transformers import AutoTokenizer, pipeline


# =========================================================
# 1. PDF TEXT EXTRACTION & CHUNKING
# =========================================================
def load_and_chunk_pdf(pdf_path: str, chunk_size: int = 600, overlap: int = 100) -> list[dict]:
    """Reads a PDF file page-by-page and chunks text with page metadata."""
    reader = PdfReader(pdf_path)
    chunks = []

    print(f"📄 Processing PDF: {pdf_path} ({len(reader.pages)} pages total)")

    for page_num, page in enumerate(reader.pages, start=1):
        text = page.extract_text()
        if not text or not text.strip():
            continue

        # Clean basic white-space formatting
        text = " ".join(text.split())

        # Sliding window text chunking
        start = 0
        while start < len(text):
            end = start + chunk_size
            chunk_text = text[start:end].strip()

            if chunk_text:
                chunks.append({
                    "text": chunk_text,
                    "page": page_num
                })
            start += (chunk_size - overlap)

    print(f"✅ Extracted {len(chunks)} searchable chunks from PDF.")
    return chunks


# =========================================================
# 2. PYTORCH PDF-RAG CLASS
# =========================================================
class PDFRAGSystem:
    def __init__(self, pdf_path: str):
        self.device = "cuda" if torch.cuda.is_available() else "cpu"
        print(f"⚙️ Using PyTorch device: {self.device}")

        # Step A: Load and Chunk PDF
        self.chunks = load_and_chunk_pdf(pdf_path)
        chunk_texts = [c["text"] for c in self.chunks]

        # Step B: Load Embedding Model & Convert Chunks to PyTorch Tensors
        print("🧠 Loading Embedding Model (all-MiniLM-L6-v2)...")
        self.embedder = SentenceTransformer("all-MiniLM-L6-v2", device=self.device)
        self.doc_embeddings = self.embedder.encode(
            chunk_texts,
            convert_to_tensor=True,
            device=self.device
        )

        # Step C: Load LLM Generator
        print("🤖 Loading LLM (TinyLlama-1.1B)...")
        model_id = "TinyLlama/TinyLlama-1.1B-Chat-v1.0"
        self.tokenizer = AutoTokenizer.from_pretrained(model_id)
        self.generator = pipeline(
            "text-generation",
            model=model_id,
            torch_dtype=torch.float16 if self.device == "cuda" else torch.float32,
            device_map="auto" if self.device == "cuda" else None,
        )
        print("✨ Pipeline Initialization Complete!\n")

    def ask(self, query: str, top_k: int = 3) -> tuple[str, list[dict]]:
        # 1. Embed user query to PyTorch tensor
        query_embedding = self.embedder.encode(query, convert_to_tensor=True, device=self.device)

        # 2. PyTorch Cosine Similarity Search
        cosine_scores = util.cos_sim(query_embedding, self.doc_embeddings)[0]
        top_results = torch.topk(cosine_scores, k=min(top_k, len(self.chunks)))

        # 3. Fetch matching PDF context chunks
        retrieved_chunks = [self.chunks[idx.item()] for idx in top_results.indices]

        context_str = "\n---\n".join(
            [f"[Page {c['page']}]: {c['text']}" for c in retrieved_chunks]
        )

        # 4. Prompt Construction
        prompt = f"""<|system|>
You are a helpful document assistant. Answer the user's question accurately using ONLY the provided context extracted from the PDF document.
Context from PDF:
{context_str}</s>
<|user|>
{query}</s>
<|assistant|>"""

        # 5. LLM Inference
        outputs = self.generator(
            prompt,
            max_new_tokens=150,
            temperature=0.1,
            do_sample=True,
            pad_token_id=self.tokenizer.eos_token_id
        )

        full_text = outputs[0]["generated_text"]
        answer = full_text.split("<|assistant|>")[-1].strip()

        return answer, retrieved_chunks


# =========================================================
# 3. RUNTIME / USAGE EXAMPLE
# =========================================================
if __name__ == "__main__":
    # Pass your PDF file path here
    PDF_FILE_PATH = "/content/HandBook_b9c6eSTUDENT HANDBOOK 2025.pdf"

    # Initialize RAG Pipeline with the PDF
    rag = PDFRAGSystem(pdf_path=PDF_FILE_PATH)

    # Query loop
    question = "What is the minimum attendance requirement for exams?"
    print(f"❓ Question: {question}\n")

    answer, retrieved_sources = rag.ask(question, top_k=2)

    print("📌 Retrieved Relevant Chunks:")
    for i, chunk in enumerate(retrieved_sources, 1):
        print(f"  [{i}] (Page {chunk['page']}): {chunk['text']}\n")

    print(f"🤖 Answer:\n{answer}")
```

```python
import torch
from google.colab import files
from pypdf import PdfReader
from sentence_transformers import SentenceTransformer, util
from transformers import AutoTokenizer, pipeline

# =========================================================
# 1. UPLOAD PDF FILE DIRECTLY IN COLAB
# =========================================================
print("📥 Please upload your PDF file below:")
uploaded = files.upload()

# Get the uploaded PDF file path
pdf_filename = list(uploaded.keys())[0]
print(f"\n✅ Uploaded File: {pdf_filename}\n")


# =========================================================
# 2. PDF TEXT EXTRACTION & CHUNKING
# =========================================================
def load_and_chunk_pdf(pdf_path: str, chunk_size: int = 600, overlap: int = 100):
    reader = PdfReader(pdf_path)
    chunks = []
    print(f"📄 Processing PDF ({len(reader.pages)} pages total)...")

    for page_num, page in enumerate(reader.pages, start=1):
        text = page.extract_text()
        if not text or not text.strip():
            continue
        text = " ".join(text.split())

        start = 0
        while start < len(text):
            end = start + chunk_size
            chunk_text = text[start:end].strip()
            if chunk_text:
                chunks.append({"text": chunk_text, "page": page_num})
            start += (chunk_size - overlap)

    print(f"✅ Created {len(chunks)} searchable chunks.")
    return chunks


# =========================================================
# 3. INITIALIZE PYTORCH MODELS
# =========================================================
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"⚙️ Running on device: {device.upper()}")

chunks = load_and_chunk_pdf(pdf_filename)
chunk_texts = [c["text"] for c in chunks]

print("🧠 Loading Embedding Model...")
embedder = SentenceTransformer("all-MiniLM-L6-v2", device=device)
doc_embeddings = embedder.encode(chunk_texts, convert_to_tensor=True, device=device)

print("🤖 Loading LLM (TinyLlama)...")
model_id = "TinyLlama/TinyLlama-1.1B-Chat-v1.0"
tokenizer = AutoTokenizer.from_pretrained(model_id)
generator = pipeline(
    "text-generation",
    model=model_id,
    torch_dtype=torch.float16 if device == "cuda" else torch.float32,
    device_map="auto" if device == "cuda" else None,
)
print("✨ Setup Complete!\n" + "=" * 50)


# =========================================================
# 4. ANSWER FUNCTION
# =========================================================
def ask_question(query: str, top_k: int = 3):
    query_embedding = embedder.encode(query, convert_to_tensor=True, device=device)
    cosine_scores = util.cos_sim(query_embedding, doc_embeddings)[0]
    top_results = torch.topk(cosine_scores, k=min(top_k, len(chunks)))

    retrieved_chunks = [chunks[idx.item()] for idx in top_results.indices]
    context_str = "\n---\n".join([f"[Page {c['page']}]: {c['text']}" for c in retrieved_chunks])

    prompt = f"""<|system|>
You are a helpful document assistant. Answer the user's question accurately using ONLY the provided context extracted from the PDF document.
Context from PDF:
{context_str}</s>
<|user|>
{query}</s>
<|assistant|>"""

    outputs = generator(
        prompt,
        max_new_tokens=150,
        temperature=0.1,
        do_sample=True,
        pad_token_id=tokenizer.eos_token_id
    )

    full_text = outputs[0]["generated_text"]
    answer = full_text.split("<|assistant|>")[-1].strip()

    return answer, retrieved_chunks


# =========================================================
# 5. INTERACTIVE QUESTION LOOP
# =========================================================
print("\n💬 You can now ask questions about your PDF!")
print("Type 'exit' or 'quit' to stop.\n")

while True:
    user_query = input("\n❓ Your Question: ")

    if user_query.strip().lower() in ["exit", "quit", "q"]:
        print("👋 Session ended.")
        break

    if not user_query.strip():
        continue

    print("🤔 Searching PDF & Generating answer...")
    answer, sources = ask_question(user_query)

    print("\n📌 Relevant Sources:")
    for idx, s in enumerate(sources, 1):
        print(f"   [{idx}] Page {s['page']}: {s['text'][:100]}...")

    print(f"\n🤖 Answer:\n{answer}\n")
    print("-" * 50)
```
